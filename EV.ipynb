```python
# Calculate Q1 and Q3
Q1 = df["battery_health_percent"].quantile(0.25)
Q3 = df["battery_health_percent"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Keep only values inside the limits
df = df[
    (df["battery_health_percent"] >= lower_limit) &
    (df["battery_health_percent"] <= upper_limit)
]

# Reset index
df.reset_index(drop=True, inplace=True)
```

```python
from sklearn.metrics import (

    accuracy_score,

    precision_score,

    recall_score,

    f1_score,

    roc_auc_score

)

results = []

for name, model in models.items():

    # Train model

    model.fit(X_train, y_train)

    # Predictions

    y_pred = model.predict(X_test)

    y_prob = model.predict_proba(X_test)[:, 1]

    # Save results

    results.append({

        "Model": name,

        "Accuracy": accuracy_score(y_test, y_pred),

        "Precision": precision_score(y_test, y_pred),

        "Recall": recall_score(y_test, y_pred),

        "F1": f1_score(y_test, y_pred),

        "ROC-AUC": roc_auc_score(y_test, y_prob)

    })

results_df = pd.DataFrame(results)

print(results_df)
```

```python
# Linear Models
from sklearn.linear_model import LogisticRegression, RidgeClassifier, SGDClassifier

# Tree Models
from sklearn.tree import DecisionTreeClassifier

# Ensemble Models
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    HistGradientBoostingClassifier,
    AdaBoostClassifier
)

# Neighbors
from sklearn.neighbors import KNeighborsClassifier

# SVM
from sklearn.svm import SVC

# Naive Bayes
from sklearn.naive_bayes import GaussianNB

# Discriminant Analysis
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)

# Neural Network
from sklearn.neural_network import MLPClassifier
```

```python
models = {

    "Logistic Regression":
        LogisticRegression(max_iter=2000),

    "Ridge Classifier":
        RidgeClassifier(),

    "SGD Classifier":
        SGDClassifier(loss="log_loss", random_state=42),

    "Decision Tree":
        DecisionTreeClassifier(random_state=42),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            random_state=42
        ),

    "Extra Trees":
        ExtraTreesClassifier(
            n_estimators=200,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(random_state=42),

    "Hist Gradient Boosting":
        HistGradientBoostingClassifier(random_state=42),

    "AdaBoost":
        AdaBoostClassifier(random_state=42),

    "KNN":
        KNeighborsClassifier(),

    "SVM":
        SVC(probability=True, random_state=42),

    "Gaussian Naive Bayes":
        GaussianNB(),

    "Linear Discriminant Analysis":
        LinearDiscriminantAnalysis(),

    "Quadratic Discriminant Analysis":
        QuadraticDiscriminantAnalysis(),

    "Neural Network":
        MLPClassifier(
            hidden_layer_sizes=(100, 50),
            max_iter=500,
            random_state=42
        )
}
```

```python
results = []

for name, model in models.items():

    # Train

    model.fit(X_train, y_train)

    # Predict classes

    y_pred = model.predict(X_test)

    # Get probability/decision score for ROC-AUC

    if hasattr(model, "predict_proba"):

        y_score = model.predict_proba(X_test)[:, 1]

    elif hasattr(model, "decision_function"):

        y_score = model.decision_function(X_test)

    else:

        y_score = None

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(y_test, y_pred, zero_division=0)

    recall = recall_score(y_test, y_pred, zero_division=0)

    f1 = f1_score(y_test, y_pred, zero_division=0)

    if y_score is not None:

        auc = roc_auc_score(y_test, y_score)

    else:

        auc = None

    results.append({

        "Model": name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1": f1,

        "ROC-AUC": auc

    })
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(

    by="F1",

    ascending=False

)

print(results_df)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

model = models["Linear Discriminant Analysis"]

y_pred = model.predict(X_test)

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Failure", "Failure"]
)

disp.plot()
plt.show()
```
